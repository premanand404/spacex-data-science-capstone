This notebook loads the official IBM teaching dataset, explores launch outcomes, and runs SQL queries. Run all cells to generate actual outputs.

```python
import pandas as pd, sqlite3, matplotlib.pyplot as plt
URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv'
df = pd.read_csv(URL)
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df['Year'] = df['Date'].dt.year
df.head()
```

```python
df.info()
df.isna().sum().sort_values(ascending=False).head(10)
```

```python
site = df.groupby('LaunchSite').agg(launches=('Class','size'), success_rate=('Class','mean')).sort_values('success_rate', ascending=False)
site
```

```python
ax = site['success_rate'].plot(kind='bar', title='Landing success rate by launch site', ylim=(0,1), ylabel='Success rate')
plt.tight_layout()
```

```python
year = df.groupby('Year').agg(launches=('Class','size'), success_rate=('Class','mean'))
year
```

```python
fig, ax = plt.subplots(figsize=(9,4))
year['success_rate'].plot(marker='o', ax=ax, ylim=(0,1), title='Success rate by year')
ax.set_ylabel('Success rate'); plt.tight_layout()
```

```python
con = sqlite3.connect(':memory:')
df.to_sql('launches', con, index=False, if_exists='replace')
query = '''SELECT LaunchSite, COUNT(*) AS launches, ROUND(AVG(Class), 3) AS success_rate FROM launches GROUP BY LaunchSite ORDER BY success_rate DESC'''
pd.read_sql_query(query, con)
```

```python
query_orbit = '''SELECT Orbit, COUNT(*) AS launches, ROUND(AVG(Class), 3) AS success_rate FROM launches GROUP BY Orbit HAVING COUNT(*) >= 3 ORDER BY success_rate DESC'''
pd.read_sql_query(query_orbit, con)
```
